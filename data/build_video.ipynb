```python
import subprocess

frame_folder = "/home/anhtv/ETC/HT21/train/HT21-01/img1"

subprocess.run([
    "ffmpeg",
    "-y",
    "-framerate", "30",
    "-i", f"{frame_folder}/%06d.jpg",
    "-c:v", "libx264",
    "-preset", "fast",
    "-tune", "zerolatency",
    "-pix_fmt", "yuv420p",
    "-profile:v", "baseline",
    "-level", "3.0",
    "output_h264.mp4"
])

print("Done! Created output_h264.mp4")
```
