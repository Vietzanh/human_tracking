```python
import cv2
import os
from ultralytics import YOLO
from tqdm import tqdm

# ===== CONFIG =====
frame_folder = "/home/anhtv/ETC/MOT15/train/ADL-Rundle-6/img1"
output_video = "output.mp4"
model_path = "/home/anhtv/ETC/Human_tracking/model/best.pt"
conf_thres = 0.3
fps = 30

# ===== LOAD MODEL =====
model = YOLO(model_path)

# ===== GET FRAME LIST =====
frames = sorted(os.listdir(frame_folder))
first_frame = cv2.imread(os.path.join(frame_folder, frames[0]))
height, width, _ = first_frame.shape

# ===== VIDEO WRITER =====
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
video_writer = cv2.VideoWriter(output_video, fourcc, fps, (width, height))

# ===== PROCESS EACH FRAME =====
for frame_name in tqdm(frames):
    frame_path = os.path.join(frame_folder, frame_name)
    frame = cv2.imread(frame_path)

    results = model.predict(frame, conf=conf_thres, verbose=False)
    annotated = results[0].plot()

    video_writer.write(annotated)

video_writer.release()
print("Video saved at:", output_video)
```

```python
cap = cv2.VideoCapture("output.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    cv2.imshow("Video", frame)

    # Nhấn q để thoát
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
